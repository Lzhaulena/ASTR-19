```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 2*np.pi, 100)
```

```python
def decaying_exp(x, a, b):
    return a * np.exp(-b * x)
```

```python
a = 2.0
b = 0.5
y = decaying_exp(x, a, b)
```

```python
plt.figure(figsize=(6,4))
plt.plot(x, y, label=f'y = {a} * exp(-{b} x)')
plt.xlabel('x')
plt.ylabel('y')
plt.ylabel('y')
plt.title('Decaying Exponential')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig('decaying-exponential.pdf')
plt.show()
```
