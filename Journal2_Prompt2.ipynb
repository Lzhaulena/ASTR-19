```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 2*np.pi, 100)
```

```python
def decaying_exp(x, a, b):
    return a * np.exp(-b * x)
```

```python
a = 2.0
b = 0.5
y_true = decaying_exp(x, a, b)
```

```python
sigma = np.linspace(2.0, 0.2, x.size)

rng = np.random.default_rng(seed=42)
noise = rng.normal(loc=0.0, scale=sigma)

y_noisy = y_true + noise
```

```python
plt.figure(figsize=(7,4))
plt.errorbar(x, y_noisy, yerr=sigma, fmt='s', color='C1', ecolor='gray', elinewidth=1, capsize=3, label='noisy data with error')
plt.plot(x, y_true, color='C0', linestyle='-', label='true model')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Decaying Exponential with heteroscedastic Noise')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig('decaying_exponential_with_errors.pdf')
plt.show()
```
